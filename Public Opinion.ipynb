```python
import ipywidgets as widgets
from IPython.display import display, HTML

# ---------------------------------------------------------------
# Candidate data (Replace photo/logo URLs with real ones)
# ---------------------------------------------------------------
candidates = [
    {
        "name": "Rigathi Gachagua",
        "party": "DCP",
        "color": "#6a11cb",
        "photo": "https://via.placeholder.com/48/6a11cb/ffffff?text=RG", # Replace with real URL
        "logo": "https://via.placeholder.com/14/6a11cb/ffffff?text=DCP" # Replace with real URL
    },
    {
        "name": "Dr. William Ruto",
        "party": "UDA",
        "color": "#0ea5e9",
        "photo": "https://via.placeholder.com/48/0ea5e9/ffffff?text=WR", # Replace with real URL
        "logo": "https://via.placeholder.com/14/0ea5e9/ffffff?text=UDA" # Replace with real URL
    },
    {
        "name": "Edwin Sifuna",
        "party": "Linda Mwananchi",
        "color": "#ec4899",
        "photo": "https://via.placeholder.com/48/ec4899/ffffff?text=ES", # Replace with real URL
        "logo": "https://via.placeholder.com/14/ec4899/ffffff?text=LM" # Replace with real URL
    },
    {
        "name": "Dr. Fred Matiang'i",
        "party": "Jubilee",
        "color": "#f59e0b",
        "photo": "https://via.placeholder.com/48/f59e0b/ffffff?text=FM", # Replace with real URL
        "logo": "https://via.placeholder.com/14/f59e0b/ffffff?text=JUB" # Replace with real URL
    },
    {
        "name": "Stephen Kalonzo Musyoka",
        "party": "Wiper",
        "color": "#10b981",
        "photo": "https://via.placeholder.com/48/10b981/ffffff?text=KM", # Replace with real URL
        "logo": "https://via.placeholder.com/14/10b981/ffffff?text=WIP" # Replace with real URL
    },
]

vote_counts = {c["name"]: 0 for c in candidates}
selected_candidate = {"name": None}

# ---------------------------------------------------------------
# Header
# ---------------------------------------------------------------
header_html = """
<div style="
    background: linear-gradient(135deg, #7c3aed 0%, #06b6d4 100%);
    padding: 24px;
    border-radius: 18px;
    box-shadow: 0 10px 30px rgba(124,58,237,0.35);
    color: white;
    font-family: 'Segoe UI', Roboto, sans-serif;
    margin-bottom: 20px;">
  <h1 style="margin:0; font-size:26px; font-weight:700; letter-spacing:-0.5px;">
    🗳️ Community Poll 2026
  </h1>
  <p style="margin:6px 0 0 0; font-size:14px; opacity:0.9;">
    Tap a candidate card to cast your vote instantly.
  </p>
</div>
"""
header = widgets.HTML(header_html)

# ---------------------------------------------------------------
# Card Rendering Function
# ---------------------------------------------------------------
def create_card_html(c, is_selected, is_disabled):
    total = sum(vote_counts.values())
    n = vote_counts[c["name"]]
    pct = (n / total * 100) if total else 0

    border = "2px solid #7c3aed" if is_selected else "1px solid #e5e7eb"
    bg = "#fdf4ff" if is_selected else "#ffffff"
    check = "✅ " if is_selected else ""

    results_html = ""
    if is_disabled:
        results_html = f"""
        <div style="text-align:right; min-width:90px;">
            <div style="font-weight:700; font-size:15px; color:#111827; margin-bottom:6px;">{pct:.1f}%</div>
            <div style="height:8px; background:#e5e7eb; border-radius:4px; overflow:hidden; width:90px;">
                <div style="width:{pct}%; height:100%; background:{c['color']}; border-radius:4px; transition: width 0.8s ease;"></div>
            </div>
            <div style="font-size:12px; color:#6b7280; margin-top:6px;">{n} vote{'s' if n != 1 else ''}</div>
        </div>
        """

    return f"""
    <div style="
        display:flex; align-items:center; justify-content:space-between; gap:14px;
        background:{bg}; border:{border};
        border-radius:14px; padding:16px 18px;
        box-shadow:0 2px 6px rgba(0,0,0,0.04);
        font-family:'Segoe UI', Roboto, sans-serif;
        margin-bottom: 12px;
        box-sizing: border-box; width: 100%;">

      <div style="display:flex; align-items:center; gap:14px; flex:1;">
          <img src="{c['photo']}" style="width:48px; height:48px; border-radius:50%; object-fit:cover; flex-shrink:0;">

          <div>
            <div style="font-weight:600; font-size:16px; color:#111827;">{check}{c['name']}</div>
            <div style="font-size:13px; color:#6b7280; display:flex; align-items:center; gap:6px; margin-top:4px;">
              <img src="{c['logo']}" style="width:14px; height:14px; border-radius:4px; object-fit:cover;">
              {c['party']}
            </div>
          </div>
      </div>

      {results_html}
    </div>
    """

# ---------------------------------------------------------------
# Interactive Logic
# ---------------------------------------------------------------
cards = []
card_outputs = []
card_buttons = []

def on_card_click(btn):
    if selected_candidate["name"] is not None:
        return
    chosen = btn.tooltip
    selected_candidate["name"] = chosen
    vote_counts[chosen] += 1
    update_ui()

def update_ui():
    is_disabled = selected_candidate["name"] is not None
    for i, c in enumerate(candidates):
        is_sel = selected_candidate["name"] == c["name"]
        card_outputs[i].value = create_card_html(c, is_sel, is_disabled)
        card_buttons[i].disabled = is_disabled

for c in candidates:
    out = widgets.Output()
    with out:
        display(HTML(create_card_html(c, False, False)))

    btn = widgets.Button(description="", tooltip=c["name"])
    btn.add_class("poll-card-btn")
    btn.on_click(on_card_click)

    stack = widgets.GridBox(
        [btn, out],
        layout=widgets.Layout(width="100%", grid_template_columns="1fr", grid_template_rows="1fr")
    )
    btn.layout.grid_area = "1 / 1"
    out.layout.grid_area = "1 / 1"
    btn.layout.margin = "0"

    cards.append(stack)
    card_outputs.append(out)
    card_buttons.append(btn)

display(HTML("""
<style>
.poll-card-btn {
    opacity: 0 !important;
    cursor: pointer;
    z-index: 10;
    width: 100% !important;
    height: 100% !important;
    border: none !important;
}
.poll-card-btn:disabled {
    cursor: not-allowed;
}
</style>
"""))

app = widgets.VBox(
    [header] + cards,
    layout=widgets.Layout(width="580px", padding="20px", background_color="#f9fafb")
)

display(app)
```
